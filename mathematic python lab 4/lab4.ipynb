```python
# e - eulers number - value = 2.718 used in physics
#e to the power of x
x = np.arange(5)
y = np.exp(x) # e**x
print(y)
```

```python
x = np.linspace(-2,2)
y = np.exp(-x)
y1 = np.exp(1.0*x)
y2 = np.exp(1.2*x)
y3 = np.exp(1.5*x)
plt.plot(x,y, label = '$y = e^{-x}$')
plt.plot(x,y1, label = '$y_1 = e^{x}$')
plt.plot(x,y2, label = '$y_2 = e^{1.2x}$')
plt.plot(x,y3, label = '$y_3 = e^{1.5x}$')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# 10*x +1
# 220.0 * log(x)
# exp(0.059+x)

x = np.array([1.5, 2.0, 5.0, 10.0, 20.0, 50.0, 80.0, 100.0])
y1 = 10*x + 1
y2 = 220.0 * np.log(x) # logarithmic function
y3 = np.exp(0.069*x)
plt.plot(x, y1, "o-", label = '$y_1 = 10x + 1$')
plt.plot(x, y2, "s-", label = '$y_2 = 220.0 \\cdot\\ log(x)$') # \\cdot\\ is used to write dot in the formular
plt.plot(x, y3, "d-", label = '$y_3 = e^{0.069x}$')
plt.grid()
plt.legend()
plt.show()
```

```python
# semilogx is used to scale so that y axis stays linear but x axis is logarithmic

plt.semilogx(x, y1, "o-", label = '$y_1 = 10x + 1$')
plt.semilogx(x, y2, "s-", label = '$y_2 = 220.0 \cdot\ log(x)$') # r'' is used to have all inside the quotation treated as raw string
plt.semilogx(x, y3, "d-", label = '$y_3 = e^{0.069x}$')
plt.grid()
plt.legend()
plt.show()
```

```python
#semilogy scale: x-axis linear and y-axis logarithmic

plt.semilogy(x, y1, "o-", label = '$y_1 = 10x + 1$')
plt.semilogy(x, y2, "s-", label = r'$y_2 = 220.0 \cdot\ log(x)$') # r'' is used to have all inside the quotation treated as raw string
plt.semilogy(x, y3, "d-", label = '$y_3 = e^{0.069x}$')
plt.grid()
plt.legend()
plt.show()
```

```python
#loglog: both axes x and y are logarithmic
plt.loglog(x,y1)
plt.loglog(x, y1, "o-", label = '$y_1 = 10x + 1$')
plt.loglog(x, y2, "s-", label = '$y_2 = 220.0 \cdot\ log(x)$') # r'' is used to have all inside the quotation treated as raw string
plt.loglog(x, y3, "d-", label = '$y_3 = e^{0.069x}$')
plt.grid()
plt.legend()
plt.show()
```
